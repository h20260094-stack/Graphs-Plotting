## Matplotlib

Introduction to graphing/plotting tools and features

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xpoints = [1,2,3,4]
ypoints = [1,4,9,16]
```

```python
# Add marker as keyword

plt.plot(xpoints, ypoints, marker = 'o')
plt.show()
```

## Type of Viz

### Scatter Plots

Analyse relationship or pattern between 2 numerical variables

```python
plt.scatter(xpoints, ypoints)
plt.show()
```

```python
# What if xpoints and ypoints have 100 datapoints each? Create new variables x_100, y_100.
x_100 = np.array(list(range(1,101)))
y_100 = x_100**2
plt.scatter(x_100, y_100)
plt.show()


```

```python
# Add y1_100 and y2_100

plt.scatter(x_100, y1_100)
plt.scatter(x_100, y2_100)
plt.show()



```

```python
# Add y1_100 and y2_100
```

```python
# What will be the output for this graph?

y1_100 = x_100**3
y2_100 = x_100**4

plt.scatter(y_100, y2_100, color = '#FF6BF8')
plt.scatter(y1_100, y1_100, color = '#B5FF6B')

plt.show()
```

```python
# Colour each dots

col_list=np.array(['red','yellow','blue','#FF6BF8'])
plt.scatter(xpoints, ypoints, color = col_list)
plt.show()
```

```python
# Use a colormap
plt.scatter(xpoints,ypoints,c=range(len(xpoints)),cmap='viridis')
plt.show()
```

```python
# Use a colorbar
plt.scatter(xpoints,ypoints,c=range(len(xpoints)),cmap='viridis')
plt.colorbar()
plt.show()
```

```python
# Reverse the colour map
plt.scatter(xpoints,ypoints,c=range(len(xpoints)),cmap='viridis_r')
plt.colorbar()
plt.show()
# Use a colormap
```

```python
# Adjust size and opacity

plt.scatter(xpoints,ypoints,c=range(len(xpoints)),cmap='viridis_r', s = 25, alpha = 0.6)  #s for size and alpha for opacity
plt.colorbar()
plt.show()
```

```python
# Change each dot sizes

plt.scatter(xpoints,ypoints,c=range(len(xpoints)),cmap='viridis_r', s = np.array(ypoints)*10, alpha = 0.5)  #s for size and alpha for opacity
plt.colorbar()
plt.show()
```

### Bar Graph

```python
plt.bar(xpoints, ypoints)
plt.show()

# How does the interpretation change?
#if we have a category on X axis then bar graph makes sense
```

```python
category = ["A", "B", "C", "D"]
plt.bar(category, ypoints)
plt.show()
```

```python
# Horizontal Bars
plt.barh(category, ypoints)
plt.show()
```

```python
# Add colour
plt.bar(category,ypoints,color="pink")
plt.show()
```

```python
# Change bar width
plt.bar(xpoints,ypoints,width=0.2,color='red')
plt.show()
```

```python
# Add bar height
plt.barh(xpoints,ypoints,height=0.2,color="pink")
plt.show()


```

### Histogram

## Scatter Plots

```python
penguins = sns.load_dataset("penguins")

sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm")

plt.show()
```

### Line plot

```python
sns.lineplot(data=penguins, x="bill_length_mm", y="bill_depth_mm")
plt.show()
```

### Relational Plot (Relplot)

```python
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm",hue='species')
plt.show()
```

### Bar Plot

```python
sns.barplot(data=penguins, x="species", y="bill_depth_mm")
plt.show()
#here the numerical values as average values
```

How to interpret this plot?

### Count Plot

```python
sns.countplot(data=penguins, x="island")
plt.show()
```

### Box Plot

```python
sns.boxplot(data=penguins, x="species", y="body_mass_g")
plt.show()
```

### Violin Plot

```python
sns.violinplot(data=penguins, x="species", y="body_mass_g")
plt.show()
```

### Heatmap

```python
# Drop NAs and calculate the correlation matrix for numeric columns
numeric_penguins = penguins.select_dtypes(include=[np.number]).dropna()
corr_matrix = numeric_penguins.corr()

# Create the heatmap correlogram
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm_r", fmt=".2f")
plt.show()
```
